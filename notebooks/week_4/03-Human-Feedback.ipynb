```python
from langsmith import Client
```

### Attach Feedback to a specific trace in LangSmith

```python
client = Client()
```

```python
RUN_ID = "019f9aa7-1474-7383-b83f-189956dbafe1"
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="feedback",
    score=0,
    feedback_source_type="api"
)
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="comment",
    value="This is a long form free text feedback left by Tara.",
    feedback_source_type="api"
)
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="comment_with_score",
    value="This is a long form free text feedback left by Tara.",
    score=1,
    feedback_source_type="api"
)
```
